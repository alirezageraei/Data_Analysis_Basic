## Making Feature Set

```python
target = "Survived"
features = [
    "Sex",
    "Pclass",
    "AgeGroup",
    "FamilyGroup",
    "Fare",
    "Embarked"]
```

```python
X = df[features]
Y = df[target]
```

```python
X.head()
```

```python
X.info()
```

## Encoding Proccess

```python
encoded_df = pd.get_dummies(
    X,
    columns=["Sex", "AgeGroup", "FamilyGroup", "Embarked"],
    drop_first=True,
    dtype="int8")
```

```python
encoded_df.info()
```

## Train-Test Split

```python
x_train, x_test, y_train, y_test = train_test_split(
    encoded_df,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y)
```

```python
print(x_train.shape)
print(x_test.shape)

print(y_train.value_counts())
print(y_test.value_counts())
```

## **Model Training**

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model
```

```python
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

`TitleGroup` was tested as an additional feature using 5-Fold Cross-Validation.
Although it slightly improved the mean accuracy (from **80.02%** to **80.69%**), it also increased the standard deviation (from **1.68%** to **2.04%**).

For this reason, `TitleGroup` was excluded from the final feature set.

# Titanic Survival Prediction — Final Summary

In this project, the Titanic dataset was explored and used to build a Logistic Regression model for predicting passenger survival.

### Data Analysis & Feature Engineering

The analysis started by investigating the relationship between survival and important passenger characteristics such as **Sex, Pclass, Age, Fare, Family Size, and Embarked**.

Several features were engineered to provide more useful information to the model:

* `AgeGroup` was created from passenger age.
* `FamilyGroup` was created from family size.
* Categorical variables were converted using one-hot encoding.
* Missing Age values were handled by filling them with the mean.

`TitleGroup` was also tested as a potential feature. Although it slightly improved the mean Cross-Validation accuracy, further analysis showed that most of its information overlapped with existing features, especially `Sex` and `AgeGroup`. Therefore, it was excluded from the final feature set.

### Model & Evaluation

A **Logistic Regression** model was trained and evaluated using both a stratified train/test split and **5-Fold Cross-Validation**.

The final model achieved approximately:

* **Accuracy:** 79.3%
* **Precision:** 74.2%
* **Recall:** 71.0%
* **F1-score:** 72.6%

The 5-Fold Cross-Validation results were:

* **Mean Accuracy:** 80.02%
* **Standard Deviation:** 1.68%

The relatively small standard deviation indicates that the model's performance was reasonably stable across different subsets of the data.

The model also performed substantially better than a simple baseline that always predicts the majority class, which would achieve approximately 61.6% accuracy.

### Final Conclusion

The project demonstrated a complete basic Machine Learning workflow:

**Data Exploration → Feature Engineering → Model Building → Cross-Validation → Evaluation → Feature Selection → Model Interpretation**

The Logistic Regression model was able to learn meaningful patterns in passenger data, particularly the strong relationship between survival and variables such as **Sex and Pclass**.

At the same time, the model is not perfect. Its errors show that passenger survival cannot be completely explained by the available features, and some individual cases remain difficult to classify.

Overall, this project provided practical experience with **Pandas, NumPy, feature engineering, categorical encoding, Logistic Regression, Cross-Validation, evaluation metrics, and model interpretation**, forming a foundation for working with more realistic datasets and Machine Learning problems.
